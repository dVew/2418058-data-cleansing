### 2.5 Data yang tidak valid

```python
# fungsi bantu: ubah durasi (m:dd atau detik) menjadi detik
def durasi_ke_detik(x):
    if pd.isna(x):
        return np.nan
    x = str(x).strip()
    if ":" in x:
        menit, detik = x.split(":")
        return int(menit) * 60 + int(detik)
    return float(x)


putar_cek = pd.to_numeric(df_mentah["jumlah_putar"].str.replace(",", ""), errors="coerce")
detik_cek = df_mentah["durasi"].apply(durasi_ke_detik)

print("Popularitas di luar rentang 0-100:")
print(df_mentah[(df_mentah["popularitas"] < 0) | (df_mentah["popularitas"] > 100)][["judul", "popularitas"]])
print()
print("Tahun rilis tidak masuk akal (di luar 1950-2026):")
print(df_mentah[(df_mentah["tahun_rilis"] < 1950) | (df_mentah["tahun_rilis"] > 2026)][["judul", "tahun_rilis"]])
print()
print("Jumlah putar negatif:")
print(df_mentah[putar_cek < 0][["judul", "jumlah_putar"]])
print()
print("Durasi 0 detik:")
print(df_mentah[detik_cek < 30][["judul", "durasi"]])
print()
print("Tanggal yang tidak ada di kalender:")
print(df_mentah[df_mentah["tanggal_rilis"] == "31/02/1999"][["judul", "tanggal_rilis"]])
```

### 2.6 Outlier

Untuk outlier saya pakai metode IQR (batas bawah = Q1 - 1.5 x IQR, batas atas = Q3 + 1.5 x IQR). Nilai yang ditandai IQR tidak langsung dibuang, saya lihat dulu satu per satu apakah memang salah atau cuma nilai yang besar tapi wajar.

```python
def batas_iqr(seri):
    q1 = seri.quantile(0.25)
    q3 = seri.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr


# hanya nilai yang masuk akal dulu (bukan negatif / 0) supaya batasnya tidak kacau
detik_valid = detik_cek[detik_cek >= 30]
bawah, atas = batas_iqr(detik_valid)
print(f"Durasi (detik): batas bawah={bawah:.0f}, batas atas={atas:.0f}")
print(df_mentah.loc[(detik_cek > atas) | ((detik_cek < bawah) & (detik_cek >= 30)), ["judul", "durasi"]])
print()

putar_valid = putar_cek[putar_cek > 0]
bawah, atas = batas_iqr(putar_valid)
print(f"Jumlah putar: batas bawah={bawah:,.0f}, batas atas={atas:,.0f}")
print(df_mentah.loc[(putar_cek > atas) | ((putar_cek < bawah) & (putar_cek > 0)), ["judul", "jumlah_putar"]])
```

Hasil pengecekan:
- **Durasi:** IQR menandai dua lagu, yaitu `58:20` (Kangen) dan `6:30` (Hotel California). Yang 58 menit jelas tidak wajar untuk satu lagu, jadi itu saya anggap salah. Hotel California memang lagu yang panjang tapi 6 menit lebih masih wajar, jadi tidak saya ubah.
- **Jumlah putar:** `95.000.000.000` (Gangnam Style) jauh di atas yang lain. Itu sepertinya salah ketik (kelebihan nol), jadi saya anggap outlier yang salah. Lagu yang jumlah putarnya besar tapi wajar tidak saya ubah.

## 3. Data Cleansing

Pertama saya salin dulu data mentahnya ke variabel `df`, supaya data asli (`df_mentah`) tetap utuh dan bisa dibandingkan nanti.

Urutan pembersihan saya atur begini: teks dirapikan dulu, lalu tipe data diperbaiki, lalu nilai tidak valid dijadikan kosong, lalu duplikat dibuang, lalu outlier ditangani, dan terakhir baru nilai kosong diisi. Duplikat dibuang sebelum pengisian supaya nilai median tidak terpengaruh baris ganda.

```python
df = df_mentah.copy()
print("Jumlah baris awal:", len(df))
```

### 3.1 Standardisasi teks (judul, artis, genre)

```python
# buang spasi di depan/belakang dan spasi ganda
for kol in ["judul", "artis", "genre"]:
    df[kol] = df[kol].str.strip().str.replace(r"\s+", " ", regex=True)


# judul/artis yang semuanya huruf besar atau semuanya kecil dirapikan jadi huruf kapital di awal kata
# (nama pendek seperti BTS, PSY, ABBA sengaja dibiarkan karena memang singkatan)
def rapikan_judul(j):
    if j.isupper() or j.islower():
        return j.title()
    return j


def rapikan_artis(a):
    if (a.isupper() and len(a) > 4) or a.islower():
        return a.title()
    return a


df["judul"] = df["judul"].apply(rapikan_judul)
df["artis"] = df["artis"].apply(rapikan_artis)

# genre: samakan penulisan lewat kamus
peta_genre = {
    "pop": "Pop",
    "rock": "Rock",
    "hip hop": "Hip-Hop",
    "hip-hop": "Hip-Hop",
    "kpop": "K-Pop",
    "k-pop": "K-Pop",
    "edm": "EDM",
    "funk": "Funk",
    "latin": "Latin",
    "disco": "Disco",
}
df["genre"] = df["genre"].str.lower().map(peta_genre)

print("Genre setelah distandarkan:")
print(df["genre"].value_counts(dropna=False))
```

### 3.2 Perbaikan tipe data dan format

```python
# jumlah_putar: hapus koma ribuan lalu jadikan angka
df["jumlah_putar"] = pd.to_numeric(df["jumlah_putar"].str.replace(",", ""), errors="coerce")

# durasi: samakan semuanya jadi detik
df["durasi_detik"] = df["durasi"].apply(durasi_ke_detik)
df = df.drop(columns="durasi")


# tanggal_rilis: coba tiga format yang ada, kalau gagal dijadikan NaT (kosong)
def parse_tanggal(teks):
    if pd.isna(teks):
        return pd.NaT
    for fmt in ["%Y-%m-%d", "%d/%m/%Y", "%d-%m-%Y"]:
        try:
            return pd.Timestamp(datetime.strptime(teks.strip(), fmt))
        except ValueError:
            continue
    return pd.NaT


df["tanggal_rilis"] = pd.to_datetime(df["tanggal_rilis"].apply(parse_tanggal))

df[["judul", "tanggal_rilis", "durasi_detik", "jumlah_putar"]].head(10)
```

### 3.3 Menangani data yang tidak valid

Nilai yang tidak mungkin saya ubah jadi kosong (NaN), lalu diisi di langkah missing value nanti.

```python
valid_awal = df[["popularitas", "tahun_rilis", "jumlah_putar", "durasi_detik"]].notna().sum()

df.loc[(df["popularitas"] < 0) | (df["popularitas"] > 100), "popularitas"] = np.nan
df.loc[(df["tahun_rilis"] < 1950) | (df["tahun_rilis"] > 2026), "tahun_rilis"] = np.nan
df.loc[df["jumlah_putar"] <= 0, "jumlah_putar"] = np.nan
df.loc[df["durasi_detik"] < 30, "durasi_detik"] = np.nan

valid_akhir = df[["popularitas", "tahun_rilis", "jumlah_putar", "durasi_detik"]].notna().sum()
print("Nilai tidak valid yang dijadikan kosong:")
print(valid_awal - valid_akhir)
```

### 3.4 Menghapus data duplikat

```python
sebelum = len(df)
df = df.drop_duplicates(subset=["judul", "artis"], keep="first").reset_index(drop=True)
print(f"Baris sebelum: {sebelum}, sesudah: {len(df)}, dibuang: {sebelum - len(df)}")
```

### 3.5 Menangani outlier

Sesuai hasil pengecekan di bagian 2.6, hanya nilai yang memang tidak masuk akal yang saya perbaiki. Durasi lebih dari 20 menit dan jumlah putar lebih dari 10 miliar saya anggap salah input, lalu dijadikan kosong supaya nanti diisi median.

```python
print("Outlier durasi:")
print(df.loc[df["durasi_detik"] > 1200, ["judul", "durasi_detik"]])
print()
print("Outlier jumlah putar:")
print(df.loc[df["jumlah_putar"] > 10_000_000_000, ["judul", "jumlah_putar"]])

df.loc[df["durasi_detik"] > 1200, "durasi_detik"] = np.nan
df.loc[df["jumlah_putar"] > 10_000_000_000, "jumlah_putar"] = np.nan
```

### 3.6 Menangani missing value

Cara yang saya pilih untuk tiap kolom:
- `tahun_rilis`: diisi dari tahun pada `tanggal_rilis` (karena dua kolom itu berhubungan). Kalau tanggal juga kosong, baru dipakai median.
- `genre`: diisi `Tidak Diketahui`, karena genre tidak bisa ditebak dengan median/rata-rata.
- `popularitas`, `jumlah_putar`, `durasi_detik`: diisi median. Saya pilih median (bukan rata-rata) karena nilainya bisa berat sebelah dan median tidak terpengaruh nilai ekstrem.
- `tanggal_rilis`: dibiarkan kosong. Tahunnya sudah ada di kolom `tahun_rilis` dan tanggal pastinya tidak dipakai di analisis, jadi tidak perlu ditebak-tebak.

```python
print("Missing value sebelum diisi:")
print(df.isna().sum()[df.isna().sum() > 0])

df["tahun_rilis"] = df["tahun_rilis"].fillna(df["tanggal_rilis"].dt.year)
df["tahun_rilis"] = df["tahun_rilis"].fillna(df["tahun_rilis"].median())
df["genre"] = df["genre"].fillna("Tidak Diketahui")

for kol in ["popularitas", "jumlah_putar", "durasi_detik"]:
    df[kol] = df[kol].fillna(df[kol].median())

# setelah tidak ada yang kosong, tipe datanya dirapikan jadi bilangan bulat
df["tahun_rilis"] = df["tahun_rilis"].astype(int)
df["popularitas"] = df["popularitas"].round().astype(int)
df["jumlah_putar"] = df["jumlah_putar"].round().astype("int64")
df["durasi_detik"] = df["durasi_detik"].round().astype(int)

print()
print("Missing value sesudah diisi:")
print(df.isna().sum()[df.isna().sum() > 0])
```

### 4.2 Membuat kolom turunan

```python
TAHUN_SEKARANG = 2026

# dekade rilis, misal 2017 -> 2010-an
df["dekade"] = (df["tahun_rilis"] // 10 * 10).astype(str) + "-an"

# usia lagu dalam tahun
df["usia_lagu"] = TAHUN_SEKARANG - df["tahun_rilis"]

# durasi dalam menit
df["durasi_menit"] = (df["durasi_detik"] / 60).round(2)

# kategori durasi: pendek (sampai 3:30), sedang (sampai 5:00), panjang (lebih dari 5:00)
df["kategori_durasi"] = pd.cut(
    df["durasi_detik"], bins=[0, 210, 300, np.inf], labels=["Pendek", "Sedang", "Panjang"]
).astype(str)

# kategori popularitas: rendah (<70), sedang (70-84), tinggi (85 ke atas)
df["kategori_popularitas"] = pd.cut(
    df["popularitas"], bins=[-1, 69, 84, 100], labels=["Rendah", "Sedang", "Tinggi"]
).astype(str)

# lagu lokal (artis Indonesia) atau internasional
df["asal_lagu"] = np.where(df["negara_artis"] == "Indonesia", "Lokal", "Internasional")

# rata-rata jumlah putar per tahun sejak rilis
df["putar_per_tahun"] = (df["jumlah_putar"] / df["usia_lagu"].clip(lower=1)).round().astype("int64")

df[["judul", "dekade", "usia_lagu", "durasi_menit", "kategori_durasi", "kategori_popularitas", "asal_lagu", "putar_per_tahun"]].head(10)
```

## 5. Perbandingan Kondisi Data Sebelum dan Sesudah

### 5.1 Tabel perbandingan

```python
def hitung_tidak_valid(d_pop, d_tahun, d_putar, d_detik):
    return int(((d_pop < 0) | (d_pop > 100)).sum()
               + ((d_tahun < 1950) | (d_tahun > 2026)).sum()
               + (d_putar <= 0).sum()
               + (d_detik < 30).sum())


tidak_valid_sebelum = hitung_tidak_valid(
    df_mentah["popularitas"], df_mentah["tahun_rilis"], putar_cek, detik_cek
) + 1  # +1 untuk tanggal 31/02/1999

outlier_sebelum = int((detik_cek > 1200).sum() + (putar_cek > 10_000_000_000).sum())

perbandingan = pd.DataFrame(
    {
        "Sebelum": {
            "Jumlah baris": len(df_mentah),
            "Jumlah kolom": df_mentah.shape[1],
            "Total missing value": int(df_mentah.isna().sum().sum()),
            "Baris duplikat": int(kunci.duplicated().sum()),
            "Variasi penulisan genre": df_mentah["genre"].nunique(),
            "Nilai tidak valid": tidak_valid_sebelum,
            "Outlier": outlier_sebelum,
            "Kolom dengan tipe data keliru": 3,  # jumlah_putar, durasi, tanggal_rilis (terbaca teks)
        },
        "Sesudah": {
            "Jumlah baris": len(df),
            "Jumlah kolom": df.shape[1],
            "Total missing value": int(df.isna().sum().sum()),
            "Baris duplikat": int(df.duplicated(subset=["judul", "artis"]).sum()),
            "Variasi penulisan genre": df["genre"].nunique(),
            "Nilai tidak valid": hitung_tidak_valid(
                df["popularitas"], df["tahun_rilis"], df["jumlah_putar"], df["durasi_detik"]
            ),
            "Outlier": int((df["durasi_detik"] > 1200).sum() + (df["jumlah_putar"] > 10_000_000_000).sum()),
            "Kolom dengan tipe data keliru": 0,  # sudah angka dan tanggal semua
        },
    }
)
perbandingan
```

Catatan:
- Pada kondisi sesudah masih ada 2 sel kosong di `tanggal_rilis` (Sunday Morning dan Sephia). Itu memang sengaja tidak diisi, alasannya sudah dijelaskan di bagian 3.6.

## 6. Kesimpulan

Teknik cleansing yang saya terapkan:
- standardisasi teks (spasi, huruf besar/kecil, dan kamus genre),

- penanganan outlier dengan metode IQR yang dicek manual,
